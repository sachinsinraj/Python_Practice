```python
def temp(temperature,unit):
    """this function convert temp between celcius and fahrenheit"""
    if unit=='C':
        return temperature * 9/5 + 32
    elif unit=='F':
        return (temperature-32)*5/9
    else:
        return None
    
print(temp(25,'C'))
    
```

```python
def strong_password(password):
    """this function check is password is strong or not"""
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
    
print(strong_password("WeakPwd"))
print(strong_password("StrongPwd"))
```

```python
def palin(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print(palin("A man a plan a canal panama"))
print(palin("Hello"))
```

```python
def fact(n):
    if n==0:
        return 1
    else:
        return n * fact(n-1)
    
print(fact(5))
```
